本 Notebook 是整理后的交互式运行入口。核心分析代码位于 `../src/problem2_analysis.py`，统一入口位于 `../src/problem2.py`。输入数据在 `../data/`，运行生成的换分表、TOPSIS 结果和图片在 `../outputs/`。

```python
from pathlib import Path
import runpy

def find_problem_dir():
    cwd = Path.cwd().resolve()
    candidates = []
    for base in [cwd, *cwd.parents]:
        candidates.extend([
            base,
            base / '问题二',
            base / '源程序代码' / '问题二',
        ])
    for candidate in candidates:
        if (candidate / 'src' / 'problem2.py').exists():
            return candidate
    raise FileNotFoundError('未找到问题二目录，请从仓库内部启动 Jupyter/VSCode。')

PROBLEM_DIR = find_problem_dir()
print(f'问题二目录：{PROBLEM_DIR}')
```

```python
runpy.run_path(str(PROBLEM_DIR / 'src' / 'problem2.py'), run_name='__main__')
```
