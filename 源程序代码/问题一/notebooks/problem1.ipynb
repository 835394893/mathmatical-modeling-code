本 Notebook 是整理后的交互式运行入口。核心分析代码统一位于 `../src/problem1.py`，输入数据在 `../data/`，运行输出在 `../outputs/`。

问题一现在直接使用新目录结构，不再依赖兼容层或额外的 `problem1_analysis.py`。

```python
from pathlib import Path
import runpy

def find_problem_dir():
    cwd = Path.cwd().resolve()
    candidates = []
    for base in [cwd, *cwd.parents]:
        candidates.extend([
            base,
            base / '问题一',
            base / '源程序代码' / '问题一',
        ])
    for candidate in candidates:
        if (candidate / 'src' / 'problem1.py').exists():
            return candidate
    raise FileNotFoundError('未找到问题一目录，请从仓库内部启动 Jupyter/VSCode。')

PROBLEM_DIR = find_problem_dir()
print(f'问题一目录：{PROBLEM_DIR}')
```

```python
runpy.run_path(str(PROBLEM_DIR / 'src' / 'problem1.py'), run_name='__main__')
```
